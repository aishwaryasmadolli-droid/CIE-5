```python
import spacy
from nltk import ngrams
nlp=spacy.load("en_core_web_sm")
text=nlp("AI is simulation of human intelligence by machine.It includes expects system NLP speech recognition and machine vision")
bigrams=ngrams(text, 2)
print("\n Bigrams :")
print(list(bigrams))
trigrams=ngrams(text, 3)
print("\n trigrams:")
print(list(trigrams))
```
